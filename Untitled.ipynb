```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
from sklearn.metrics.pairwise import cosine_similarity
import torch


documents = {
    "Indian Classical Music": [
        """
        Indian Classical Music is a traditional musical system that has
        developed over many centuries. It is mainly based on the concepts
        of raga and tala. A raga provides the melodic structure of a musical
        performance, while tala provides its rhythmic framework.
        """,

        """
        Indian Classical Music has two major traditions known as Hindustani
        and Carnatic music. Hindustani music is mainly associated with
        northern India, while Carnatic music is mainly practiced in southern
        India. Both traditions include vocal music and instrumental
        performances.
        """
    ],

    "Western Classical Music": [
        """
        Western Classical Music is a tradition of art music that developed
        mainly in Europe. It contains several historical periods including
        the Baroque, Classical, Romantic, and Modern periods. Musical notation,
        harmony, melody, and orchestration are important parts of this tradition.
        """,

        """
        Symphonies, concertos, sonatas, and operas are important forms of
        Western Classical Music. Famous composers include Mozart, Beethoven,
        Bach, and Vivaldi. Large orchestras commonly contain string,
        woodwind, brass, and percussion instruments.
        """
    ],

    "Music Theory": [
        """
        Music Theory explains the basic principles used to understand and
        construct music. It includes concepts such as notes, scales, chords,
        melody, harmony, rhythm, tempo, and musical notation.
        """,

        """
        A melody is a sequence of musical notes that forms a recognizable
        musical idea. Harmony describes the combination of different notes
        and chords, while rhythm deals with the timing and duration of sounds.
        Scales organize musical notes according to specific patterns.
        """
    ],

    "Musical Instruments": [
        """
        Musical instruments are objects designed to produce musical sounds.
        They can be classified into different families such as string,
        wind, percussion, and keyboard instruments. Each family produces
        sound using a different physical mechanism.
        """,

        """
        A violin produces sound through vibrating strings, while instruments
        such as flutes produce sound through vibrating air. Percussion
        instruments produce sound through striking or shaking. A piano uses
        a keyboard mechanism to cause hammers to strike strings.
        """
    ],

    "Modern Music Technology": [
        """
        Modern music production uses digital technology for recording,
        editing, mixing, and mastering audio. Digital Audio Workstations
        allow musicians and producers to record multiple tracks and combine
        them into complete musical compositions.
        """,

        """
        The Internet has changed how music is distributed and consumed.
        Streaming services allow listeners to access large music libraries.
        Artificial intelligence and machine learning are also being used for
        music recommendation, audio analysis, and automatic music generation.
        """
    ]
}


document_names = list(documents.keys())

document_texts = []

for name in document_names:
    text = " ".join(documents[name])
    document_texts.append(text)


print("Loading embedding model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)


document_embeddings = embedding_model.encode(
    document_texts,
    convert_to_numpy=True
)


print("Loading Question Answering model...")

qa_tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-cased-distilled-squad"
)

qa_model = AutoModelForQuestionAnswering.from_pretrained(
    "distilbert-base-cased-distilled-squad"
)


question = input("\nEnter your Music question: ")


question_embedding = embedding_model.encode(
    [question],
    convert_to_numpy=True
)


similarities = cosine_similarity(
    question_embedding,
    document_embeddings
)[0]


best_index = similarities.argmax()

best_document = document_names[best_index]

best_score = similarities[best_index]

best_context = document_texts[best_index]


print("\nSEMANTIC SEARCH RESULT\n")

print("Question:", question)

print("Most Relevant Document:", best_document)

print("Cosine Similarity:", round(float(best_score), 4))


print("\nDocument Similarity Scores:")

for i in range(len(document_names)):
    print(
        document_names[i],
        "->",
        round(float(similarities[i]), 4)
    )


inputs = qa_tokenizer(
    question,
    best_context,
    return_tensors="pt",
    truncation=True
)


with torch.no_grad():
    outputs = qa_model(**inputs)


start_scores = outputs.start_logits
end_scores = outputs.end_logits

start_index = torch.argmax(start_scores, dim=1).item()
end_index = torch.argmax(end_scores, dim=1).item()


if end_index < start_index:
    end_index = start_index


answer_tokens = inputs["input_ids"][0][
    start_index:end_index + 1
]

answer = qa_tokenizer.decode(
    answer_tokens,
    skip_special_tokens=True
)


start_probability = torch.softmax(
    start_scores, dim=1
)[0][start_index]

end_probability = torch.softmax(
    end_scores, dim=1
)[0][end_index]

confidence = start_probability * end_probability


print("\nEXTRACTIVE QUESTION ANSWERING\n")

print("Answer:", answer)

print("Confidence:", round(float(confidence), 4))
```
