```python
import spacy
nlp = spacy.load("en_core_web_sm")
text= "This is good day in jaipur You mistakenly passed a comparison statement as the first argument to displacy.render. Instead, you need to pass the doc object itself as the first argument, and specify the visualization type using the style parameter which accepts strings like dep for dependencies or ent for entities."
doc = nlp(text)

for ent in doc.ents:
    print(ent.text,ent.start_char,ent.end_char,ent.label_)

from spacy import displacy
displacy.render(doc,style="ent" , jupyter = True)
    
```
