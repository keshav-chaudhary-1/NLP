```python
# Classical Method
import spacy
nlp = spacy.load("en_core_web_sm")
text = "This is a good day. Isn't it's a good thing?"
doc = nlp(text)
for tokens in doc:
    print(tokens , tokens.is_punct,tokens.is_alpha)
```

```python
#Modern Method using bert base uncased model which works on WordPiece.
from transformers import AutoTokenizer
tokenizer  = AutoTokenizer.from_pretrained("bert-base-uncased")
text = "This is a good day. Isn't it's a good thing?,Atrillionrainbowspoppped is a new word" #
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
#this method using openai-community/gpt2 model which works on Byte-Pair Encoding (BPE)
from transformers import AutoTokenizer
tokenizer  = AutoTokenizer.from_pretrained("openai-community/gpt2")
text = "This is a good day. Isn't it's a good thing?,Atrillionrainbowspoppped is a new word" #
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
#this method using t5-small model which works on (SentencePiece (BPE)) 
from transformers import AutoTokenizer
tokenizer  = AutoTokenizer.from_pretrained("t5-small")
text = "This is a good day. Isn't it's a good thing?,Atrillionrainbowspoppped is a new word" #
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
# Classical Method of Checking stop words
import spacy
nlp = spacy.load("en_core_web_sm")
text = "This is a good day. Isn't it's a good thing?"
doc = nlp(text)
for tokens in doc:
    print(tokens ,tokens.is_stop)
```

```python
# Classical Method of Checking and printing stop_words.
import spacy
nlp = spacy.load("en_core_web_sm")
print(len(nlp.Defaults.stop_words)) #len function to print length of stop words from en_core_web_sm dataset.
print(nlp.Defaults.stop_words) # this method to print all the stop word from en_core_wen_wm dataset.
```
