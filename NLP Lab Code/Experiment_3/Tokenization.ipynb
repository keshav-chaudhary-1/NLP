```python
# Tokenization using spaCy
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

text = "Natural Language Processing is an important field of Artificial Intelligence."

# Process text
doc = nlp(text)

print("Original Text:")
print(text)

print("\nTokens:")
for token in doc:
    print(token.text)
```

```python
# Tokenization using BERT, GPT, T5 and DeepSeek
from transformers import AutoTokenizer

text = "Natural Language Processing is amazing!"

models = {
    "BERT": "bert-base-uncased",
    "GPT-2": "gpt2",
    "T5": "t5-small",
    "DeepSeek": "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"
}

for model_name, model_id in models.items():

    print("\n" + "=" * 50)
    print(model_name)
    print("=" * 50)

    tokenizer = AutoTokenizer.from_pretrained(model_id)

    tokens = tokenizer.tokenize(text)

    print("Tokens:")
    print(tokens)

    print("\nToken IDs:")
    print(tokenizer.convert_tokens_to_ids(tokens))
```

```python
#Extract and read text from pdf and docx.
import spacy
from pypdf import PdfReader
from docx import Document

nlp = spacy.load("en_core_web_sm")

reader = PdfReader("sample.pdf")
document = Document("sample.docx")

text = ""
docx_text =""

#for pdf extracted text.....
for page in reader.pages:
    page_text = page.extract_text()

    if page_text:
        text += page_text + ","

print("Extracted Text from PDF:")
print(text)

doc = nlp(text) 

print("\nTokens from PDF:")
for token in doc:
    print(token.text)
    
#for document extracted text......
for paragraph in document.paragraphs:
    docx_text += paragraph.text + ","
    
print("Extracted Text from word:")
print(docx_text)

docx_doc = nlp(docx_text)

print("\nTokens from Word:")
for token in docx_doc:
    print(token.text)
```

```python
# Remove Stopwords using spaCy
import spacy

nlp = spacy.load("en_core_web_sm")

text = input("Enter a paragraph: ")

doc = nlp(text)

print("\nOriginal Paragraph:")
print(text)

print("\nAfter Removing Stopwords:")

result = []

for token in doc:
    if not token.is_stop:
        result.append(token.text)

print(" ".join(result))
```

```python
# We can also remove punctuation along with stopwords:(Extra Implemented part)
import spacy

nlp = spacy.load("en_core_web_sm")

text = input("Enter a paragraph: ")

doc = nlp(text)

result = []

for token in doc:
    if not token.is_stop and not token.is_punct:
        result.append(token.text)

print("\nAfter Removing Stopwords and Punctuation:")
print(" ".join(result))
```

```python
# Test spaCy Token Attributes

import spacy

nlp = spacy.load("en_core_web_sm")

text = "Hello Keshav!"

doc = nlp(text)

for token in doc:

    print("=" * 50)

    print("Token       :", token.text)
    print("lower_      :", token.lower_)
    print("shape_      :", token.shape_)
    print("prefix_     :", token.prefix_)
    print("suffix_     :", token.suffix_)

    print("is_alpha    :", token.is_alpha)
    print("is_digit    :", token.is_digit)
    print("is_punct    :", token.is_punct)
    print("is_space    :", token.is_space)
    print("is_title    :", token.is_title)

    print("like_email  :", token.like_email)
    print("like_url    :", token.like_url)

    print("i           :", token.i)
    print("idx         :", token.idx)

    print("n_lefts     :", token.n_lefts)
    print("n_rights    :", token.n_rights)
```
