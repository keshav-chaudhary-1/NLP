```python
import os
import spacy
from pypdf import PdfReader
from docx import Document
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# READ TEXT FROM FILE
def read_file(file_path):

    extension = os.path.splitext(file_path)[1].lower()
    
    # Read PDF
    if extension == ".pdf":
        reader = PdfReader(file_path)
        text = ""
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
        return text

    # Read DOCX
    elif extension == ".docx":
        document = Document(file_path)
        text = ""
        for paragraph in document.paragraphs:
            text += paragraph.text + "\n"
        return text


    # Read TXT
    elif extension == ".txt":
        with open(file_path, "r", encoding="utf-8") as file:
            return file.read()
            
    # Ivalid file format        
    else:
        raise ValueError(
            "Unsupported file format. Use PDF, DOCX or TXT."
        )

# PREPROCESS TEXT
def preprocess(text):
    doc = nlp(text)
    tokens = []
    for token in doc:
        if not token.is_stop and not token.is_punct and token.is_alpha:
            tokens.append(token.lemma_.lower())
    return tokens

# TAKE TWO FILES AS INPUT
file1 = "sample_document_1.txt"
file2 = "sample_document_2.txt"

# EXTRACT TEXT using "read_file" user defined function 
text1 = read_file(file1)
text2 = read_file(file2)

print("TEXT FROM FIRST FILE:-")
print(text1)
print("\nTEXT FROM SECOND FILE:-")
print(text2)

# PREPROCESS BOTH DOCUMENTS using "preprocess" user defined function 
tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

processed_text1 = " ".join(tokens1)
processed_text2 = " ".join(tokens2)

# TF-IDF VECTORIZATION
documents = [
    processed_text1,
    processed_text2
]
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

# COSINE SIMILARITY
similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
score = similarity[0][0]
print("\nCOSINE SIMILARITY:-")
print("Similarity Score:", score)

# COMMON TOKENS
common_tokens = sorted(set(tokens1).intersection(set(tokens2)))
print("\nCOMMON TOKENS:-")
if common_tokens:
    for token in common_tokens:
        print(token)
else:
    print("No common tokens found.")
```
