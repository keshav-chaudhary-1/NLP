```python
import spacy

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# TAKE SENTENCE AS INPUT
text = "Natural language processing is an important field of artificial intelligence and machine learning."

# TOKENIZATION
doc = nlp(text)
tokens = []
for token in doc:
    if not token.is_punct and not token.is_space:
        tokens.append(token.text)

print("\nOriginal Sentence:-")
print(text)

print("\nTokens:-")
print(tokens)

# GENERATE N-GRAMS
number_of_tokens = len(tokens)

print("\nALL N-GRAMS:-")
# Generate 1-gram, 2-gram, 3-gram ... up to

# the total number of tokens
for n in range(1, number_of_tokens + 1):
    print(f"{n}-GRAMS:-")
    for i in range(number_of_tokens - n + 1):
        ngram = tokens[i:i + n]
        print(" ".join(ngram))
    print("\n")
```
