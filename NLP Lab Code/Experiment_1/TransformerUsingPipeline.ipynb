```python
from transformers import pipeline

def sentiment_analysis():
    classifier = pipeline("sentiment-analysis")
    text = input("Enter text: ")
    result = classifier(text)
    print("\nResult:", result)

def text_generation():
    generator = pipeline("text-generation", model="gpt2")
    text = input("Enter prompt: ")
    result = generator(
        text,
        max_new_tokens=50,
        temperature=0.7
    )
    print("\nGenerated Text:")
    print(result[0]["generated_text"])

def summarization():
    summarizer = pipeline("summarization")
    text = input("Enter long text:\n")
    result = summarizer(text, max_length=50, min_length=20, do_sample=False)
    print("\nSummary:")
    print(result[0]["summary_text"])

def question_answering():
    qa = pipeline("question-answering")
    context = input("Enter Context:\n")
    question = input("Enter Question: ")
    result = qa(question=question, context=context)
    print("\nAnswer:", result["answer"])

def translation():
    translator = pipeline(
        "translation",
        model="Helsinki-NLP/opus-mt-en-hi"
    )
    text = input("Enter English text: ")
    result = translator(text)
    print("\nHindi Translation:")
    print(result[0]["translation_text"])

while True:
    print("\n====== MENU ======")
    print("1. Sentiment Analysis")
    print("2. Text Generation")
    print("3. Summarization")
    print("4. Question Answering")
    print("5. Translation")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        sentiment_analysis()
    elif choice == "2":
        text_generation()
    elif choice == "3":
        summarization()
    elif choice == "4":
        question_answering()
    elif choice == "5":
        translation()
    elif choice == "6":
        print("Exiting...")
        break
    else:
        print("Invalid choice!")
```
